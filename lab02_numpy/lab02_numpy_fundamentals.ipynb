```python
""" QUESTIONS
- What np.zeros() does: Generates an array of specified dimensions filled entirely with floating-point 0. values.   
- What np.ones() does: Generates an array of specified dimensions filled entirely with floating-point 1. values.   
- Difference between arange() and linspace():np.arange(start, stop, step) creates values based on a given step size up to (but excluding) the stop bound. np.linspace(start, stop, num) creates a fixed number of points (num) spaced evenly across an interval, including both the start and stop values.
   
Statistical Applications:
- np.zeros(): Pre-allocating empty memory buffers to store iteration results during Monte Carlo simulations.
- np.ones(): Constructing design matrices in regression analysis where a column of ones accounts for the intercept parameter.
- np.arange(): Creating discrete integer indices, iteration steps, or domain grids for discrete random variables.
- np.linspace(): Generating a smooth continuous axis grid to plot probability density functions (PDFs) or cumulative distribution functions (CDFs) """
```

```python
# PART E
data = np.array([
    [45, 67, 82],
    [56, 91, 73],
    [64, 88, 76]
])

print("Shape:", data.shape)
print("Dimensions:", data.ndim)


print("First row:", data[0, :])
print("Second column:", data[:, 1])
print("Value 91:", data[1, 1])
print("the First two rows:\n", data[:2, :])
print("Last two columns:\n", data[:, -2:])
```

```python
# PART F

subject_means = np.mean(data, axis=0)
print("Subject Means (Math, Stats, Prog):", subject_means)


student_means = np.mean(data, axis=1)
print("Student Means (Student 1, 2, 3):", student_means)
```

```python
""" axis=0 vs axis=1
- axis=0: Collapse operations vertically down columns (collapsing rows). Yields summary measures for each variable/subject across all observations.
- axis=1: Collapse operations horizontally across rows (collapsing columns). Yields summary measures for each observation/student across all variables. """
```

```python
# LAB 2 CHALLENGE
import numpy as np
import matplotlib.pyplot as plt

# Generate 1,000 observations from N(mean=50, sd=10)
np.random.seed(42)
x = np.random.normal(loc=50, scale=10, size=1000)

# Calculate key statistics
print(f"Mean: {np.mean(x):.2f}")
print(f"Median: {np.median(x):.2f}")
print(f"Standard Deviation: {np.std(x):.2f}")
print(f"Minimum: {np.min(x):.2f}")
print(f"Maximum: {np.max(x):.2f}")

# Plot Histogram
plt.figure(figsize=(8, 5))
plt.hist(x, bins=30, edgecolor='black', alpha=0.7)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Distribution of Simulated Data")
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
