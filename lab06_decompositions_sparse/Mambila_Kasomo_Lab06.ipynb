```python
import numpy as np
from scipy import sparse

# Q1: Cholesky Decomposition
A_pd = np.array([[4, 12, -16], [12, 37, -43], [-16, -43, 98]])
L = np.linalg.cholesky(A_pd)
print("Q1 Cholesky L matrix:\n", L)

# Q2: QR Decomposition
A_qr = np.array([[1, 2], [3, 4], [5, 6]])
Q, R = np.linalg.qr(A_qr)
print("Q2 Q matrix:\n", Q)
print("Q2 R matrix:\n", R)

# Q3: Reconstruct Matrix using SVD
A_svd = np.array([[3, 1, 1], [-1, 3, 1]])
U, S, Vt = np.linalg.svd(A_svd)
Sigma = np.zeros(A_svd.shape)
Sigma[:min(A_svd.shape), :min(A_svd.shape)] = np.diag(S)
A_reconstructed = U @ Sigma @ Vt
print("Q3 Reconstructed SVD Matrix:\n", A_reconstructed)

# Q4: Condition Number
cond_num = np.linalg.cond(A_pd)
print(f"Q4 Condition Number: {cond_num:.4f}")

# Q5: Create Sparse 1000x1000 Matrix
S_sparse = sparse.random(1000, 1000, density=0.01, format='csr')
print(f"Q5 Sparse matrix shape: {S_sparse.shape}, non-zero elements: {S_sparse.nnz}")

# Q6: Why Sparse Matrices are useful
"""
Sparse matrices store only non-zero values alongside their index coordinates, saving massive amounts of RAM and allowing matrix operations to skip zero-value arithmetic.
"""
```
