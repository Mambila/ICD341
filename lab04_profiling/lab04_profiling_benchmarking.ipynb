```python
import numpy as np

# Generate 1,000,000 elements
x = np.arange(1_000_000)

# Benchmark squaring operation
%timeit x**2
```

```python
def square_loop(x):
    result = []
    for value in x:
        result.append(value ** 2)
    return result

%timeit square_loop(x)
```

```python
def square_numpy(x):
    return x ** 2

%timeit square_numpy(x)
```

```python
""" Results table
Method                             Time
Python loop                        171m/s
Numpy vectorization                3.43m/s """
```

```python
# PART D
data = np.random.normal(size=1_000_000)

%timeit np.mean(data)
%timeit np.std(data)
%timeit np.sum(data)
```
