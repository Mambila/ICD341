```python
import numpy as np

# Question 1: Solve 3x + 2y = 12 and x + 4y = 10
A1 = np.array([[3, 2], 
              [1, 4]])
b1 = np.array([12, 10])
sol1 = np.linalg.solve(A1, b1)
print(f"Q1 Solution: x = {sol1[0]:.2f}, y = {sol1[1]:.2f}")

# Question 2: Custom 3x3 System
A2 = np.array([[2, 1, 1],
              [1, 3, 2],
              [1, 0, 0]])
b2 = np.array([4, 5, 6])
sol2 = np.linalg.solve(A2, b2)
print("Q2 Solution:", sol2)

"""
Question 3: Why np.linalg.solve(A, b) is preferred over $A^{-1}b$
-Numerical Stability: Computing $A^{-1}$ directly amplifies round-off errors, particularly when $A$ is ill-conditioned (near-singular).
-Computational Efficiency: np.linalg.solve uses LU decomposition ($O(\frac{2}{3}n^3)$ operations) which is significantly faster and requires less memory than explicit matrix inversion ($O(n^3)$ operations). """
```

```python
X_raw = np.array([30, 45, 60, 100, 150, 200, 240, 250, 300, 320])
Y = np.array([25, 20, 40, 65, 120, 190, 200, 200, 260, 220])

# (i) Design Matrix
X_design = np.column_stack((np.ones_like(X_raw), X_raw))

# Estimate coefficients beta = (X^T X)^(-1) X^T Y using solve
beta = np.linalg.solve(X_design.T @ X_design, X_design.T @ Y)
intercept, slope = beta[0], beta[1]
print(f"Intercept (beta0): {intercept:.4f}")
print(f"Slope (beta1): {slope:.4f}")

# (ii) Estimate consumption for X = 181
x_new = 181
y_pred = intercept + slope * x_new
print(f"Estimated Consumption for Disposable Income of 181: {y_pred:.2f}")
```
