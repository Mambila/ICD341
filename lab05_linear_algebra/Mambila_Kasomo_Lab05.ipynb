```python
import numpy as np

# Q1: Solve 3x + 2y = 12 and x + 4y = 10
A1 = np.array([[3, 2], [1, 4]])
b1 = np.array([12, 10])
sol1 = np.linalg.solve(A1, b1)
print(f"Q1 Solution: x = {sol1[0]:.2f}, y = {sol1[1]:.2f}")

# Q2: 3x3 System using np.linalg.solve()
A2 = np.array([[2, 1, 1], [1, 3, 2], [1, 0, 0]])
b2 = np.array([4, 5, 6])
sol2 = np.linalg.solve(A2, b2)
print("Q2 Solution:", sol2)

# Q3: Why np.linalg.solve(A, b) is preferred over A^-1 * b
"""
np.linalg.solve(A, b) uses LU decomposition, which is numerically stable and has an O(2/3 n^3) time complexity. 
Explicitly calculating A^-1 is computationally expensive (O(n^3)) and introduces severe floating-point round-off errors when A is ill-conditioned.
"""

# Q4: Consumption Expenditure and Disposable Income
X_raw = np.array([30, 45, 60, 100, 150, 200, 240, 250, 300, 320])
Y = np.array([25, 20, 40, 65, 120, 190, 200, 200, 260, 220])

# (i) Design Matrix X = [1, x]
X_design = np.column_stack((np.ones_like(X_raw), X_raw))
beta = np.linalg.solve(X_design.T @ X_design, X_design.T @ Y)
beta0, beta1 = beta[0], beta[1]
print(f"Q4(i) Coefficients: Intercept = {beta0:.4f}, Slope = {beta1:.4f}")

# (ii) Estimate consumption for Disposable Income X = 181
x_new = 181
y_pred = beta0 + beta1 * x_new
print(f"Q4(ii) Estimated Consumption for X=181: {y_pred:.2f}")
```
