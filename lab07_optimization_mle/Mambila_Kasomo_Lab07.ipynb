```python
import numpy as np
from scipy.optimize import minimize
from scipy.stats import norm

# Q1: Gradient Descent for f(x) = (x - 10)^2
x = 0.0
lr = 0.1
for _ in range(50):
    grad = 2 * (x - 10)
    x = x - lr * grad
print(f"Q1 Gradient Descent Minimum x: {x:.4f}")

# Q2: Compare BFGS and CG on f(x, y) = (x - 4)^2 + (y - 6)^2
def f(p):
    return (p[0] - 4)**2 + (p[1] - 6)**2

res_bfgs = minimize(f, [0, 0], method='BFGS')
res_cg = minimize(f, [0, 0], method='CG')
print("Q2 BFGS Result:", res_bfgs.x, "Evaluations:", res_bfgs.nfev)
print("Q2 CG Result:", res_cg.x, "Evaluations:", res_cg.nfev)

# Q3: Effect of Learning Rate
"""
High learning rate: Steps overshoot the minimum, causing divergence or oscillation.
Low learning rate: Convergence is extremely slow and susceptible to getting trapped in local minima.
"""

# Q4: MLE for Mean of Normally Distributed Observations
np.random.seed(42)
data = np.random.normal(loc=15.0, scale=3.0, size=100)

def neg_log_likelihood(params, data):
    mu, sigma = params
    if sigma <= 0: return np.inf
    return -np.sum(norm.logpdf(data, loc=mu, scale=sigma))

res_mle = minimize(neg_log_likelihood, [1.0, 1.0], args=(data,), method='Nelder-Mead')
print("Q4 MLE Estimated Mean (Mu):", res_mle.x[0])

# Q5: Sigmoid Function
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# Q6: Simple Logistic Regression via MLE
X_stud = np.array([1, 2, 3, 4, 5, 6, 7, 8])
Y_pass = np.array([0, 0, 0, 0, 1, 1, 1, 1])

def logistic_nll(beta, X, Y):
    beta0, beta1 = beta
    p = sigmoid(beta0 + beta1 * X)
    p = np.clip(p, 1e-15, 1 - 1e-15)
    return -np.sum(Y * np.log(p) + (1 - Y) * np.log(1 - p))

res_logit = minimize(logistic_nll, [0.0, 0.0], args=(X_stud, Y_pass), method='BFGS')
print(f"Q6 Logistic Regression Intercept: {res_logit.x[0]:.4f}, Slope: {res_logit.x[1]:.4f}")
```
