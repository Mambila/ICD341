```python
import numpy as np
from scipy.stats import norm

# Q1: 10,000 observations from N(50, 10^2)
norm_obs = np.random.normal(loc=50, scale=10, size=10000)

# Q2: 10,000 Bernoulli observations with p = 0.3
bernoulli_obs = np.random.binomial(n=1, p=0.3, size=10000)
print(f"Q2 Bernoulli Sample Proportion: {np.mean(bernoulli_obs):.4f}")

# Q3: Monte Carlo Integration of integral x^3 dx from 0 to 1
u = np.random.uniform(0, 1, 1_000_000)
integral_est = np.mean(u**3)
print(f"Q3 Monte Carlo Integral Estimate: {integral_est:.4f}")

# Q4: Estimate Pi using 1,000,000 points
x_p, y_p = np.random.uniform(-1, 1, 1_000_000), np.random.uniform(-1, 1, 1_000_000)
pi_est = 4 * np.mean(x_p**2 + y_p**2 <= 1)
print(f"Q4 Estimated Pi: {pi_est:.4f}")

# Q5: Estimate P(X > 1.96) for X ~ N(0,1)
z_samples = np.random.normal(0, 1, 1_000_000)
prob_est = np.mean(z_samples > 1.96)
print(f"Q5 Estimated P(X > 1.96): {prob_est:.4f}")

# Q6: Importance Sampling for E(X^2) where X ~ U(0,1) using proposal g(x) = 2x
y = np.sqrt(np.random.uniform(0, 1, 1_000_000))
weights = 1.0 / (2 * y)
importance_est = np.mean((y**2) * weights)
print(f"Q6 Importance Sampling Estimate: {importance_est:.4f}")

# Q7: Difference between Importance Sampling and Control Variates
"""
Importance Sampling samples from an alternative proposal distribution g(x) to focus sampling in regions of interest and adjusts using weight ratios f(x)/g(x).
Control Variates samples from the target distribution f(x) and uses a secondary known random variable correlated with the target variable to reduce overall variance.
"""
```
