```python
import numpy as np

rng = np.random.default_rng(42)
data = np.array([3, 4, 5, 7, 8, 9])
B = 10_000

# Resample with replacement B times
boot_samples = rng.choice(data, size=(B, len(data)), replace=True)
boot_means = boot_samples.mean(axis=1)

# 95% percentile confidence interval
ci_mean = np.percentile(boot_means, [2.5, 97.5])
print(f"Sample Mean: {data.mean():.4f}")
print(f"Bootstrap Mean: {boot_means.mean():.4f}")
print(f"95% Percentile CI for Mean: [{ci_mean[0]:.4f}, {ci_mean[1]:.4f}]")
```

```python
boot_medians = np.median(boot_samples, axis=1)
ci_median = np.percentile(boot_medians, [2.5, 97.5])

print(f"Sample Median: {np.median(data):.4f}")
print(f"Bootstrap Median Mean: {boot_medians.mean():.4f}")
print(f"95% Percentile CI for Median: [{ci_median[0]:.4f}, {ci_median[1]:.4f}]")
```

```python
mu_hat = data.mean()
sigma_hat = data.std(ddof=1)

# Draw B samples of size n from Normal(mu_hat, sigma_hat)
param_samples = rng.normal(loc=mu_hat, scale=sigma_hat, size=(B, len(data)))
param_means = param_samples.mean(axis=1)
ci_param = np.percentile(param_means, [2.5, 97.5])

print(f"Fitted Normal: mu={mu_hat:.4f}, sigma={sigma_hat:.4f}")
print(f"Parametric Bootstrap 95% CI for Mean: [{ci_param[0]:.4f}, {ci_param[1]:.4f}]")
```
