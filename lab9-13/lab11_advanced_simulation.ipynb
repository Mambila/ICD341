```python
import numpy as np

rng = np.random.default_rng(42)
lam = 3.0
u = rng.uniform(0, 1, size=5000)
exp_samples = -np.log(1 - u) / lam

print(f"Sample Mean: {exp_samples.mean():.4f} (Theoretical: {1/lam:.4f})")
print(f"Sample Std:  {exp_samples.std():.4f} (Theoretical: {1/lam:.4f})")
```

```python
probs = np.array([0.1, 0.6, 0.3])
cdf = np.cumsum(probs)  # [0.1, 0.7, 1.0]

u = rng.uniform(0, 1, size=10_000)
# Map each uniform draw to category 0, 1, or 2
cat_samples = np.searchsorted(cdf, u)

values, counts = np.unique(cat_samples, return_counts=True)
for v, c in zip(values, counts):
    print(f"Category {v}: empirical prob = {c/len(cat_samples):.4f}")
```

```python
def rejection_sample_beta22(n_samples=5000, seed=42):
    rng = np.random.default_rng(seed)
    accepted = []
    trials = 0
    M = 1.5

    while len(accepted) < n_samples:
        x = rng.uniform(0, 1)
        u = rng.uniform(0, 1)
        trials += 1
        f_x = 6 * x * (1 - x)
        if u <= f_x / M:
            accepted.append(x)

    return np.array(accepted), n_samples / trials

beta_samples, acc_rate = rejection_sample_beta22(5000)
print(f"Acceptance Rate: {acc_rate:.4f} (Theoretical 1/M: {1/1.5:.4f})")
print(f"Sample Mean: {beta_samples.mean():.4f} (Theoretical: 0.5000)")
```
