```python
import numpy as np
import time

x = np.linspace(0, 10, 1_000_000)

start = time.perf_counter()
y = x**2 + 2 * x + 1
elapsed = time.perf_counter() - start

print(f"Vectorized result shape: {y.shape}, Time: {elapsed:.6f} s")
```

```python
from numba import njit
import numpy as np

@njit
def sum_of_squares(arr):
    total = 0.0
    for i in range(len(arr)):
        total += arr[i] * arr[i]
    return total

arr = np.arange(1_000_000, dtype=np.float64)

# Warm-up call (triggers JIT compilation)
_ = sum_of_squares(arr)

# Timed call (compiled machine code)
start = time.perf_counter()
result = sum_of_squares(arr)
elapsed = time.perf_counter() - start

print(f"Sum of squares: {result:.2e}, Time: {elapsed:.6f} s")
```

```python
from joblib import Parallel, delayed

data = np.array([3, 4, 5, 7, 8, 9])

def bootstrap_chunk(data, n_resamples, seed):
    rng = np.random.default_rng(seed)
    samples = rng.choice(data, size=(n_resamples, len(data)), replace=True)
    return samples.mean(axis=1)

# Run 4 parallel workers, 2,500 resamples each (10,000 total)
chunks = Parallel(n_jobs=-1)(
    delayed(bootstrap_chunk)(data, 2500, seed=100 + i) for i in range(4)
)
boot_means_parallel = np.concatenate(chunks)

print(f"Total bootstrap means: {len(boot_means_parallel)}")
print(f"Mean: {boot_means_parallel.mean():.4f}")
```

```python
import dask.array as da

# Create a 10,000 x 10,000 random array split into 2,000 x 2,000 chunks
large_arr = da.random.normal(loc=5.0, scale=2.0, size=(10_000, 10_000), chunks=(2000, 2000))

# Build lazy computation graph and execute with .compute()
mean_val = large_arr.mean().compute()
print(f"Dask Array Mean: {mean_val:.4f}")
```
